```python
import pandas as pd

# Ensure 'date' columns are datetime objects
df_pop['date'] = pd.to_datetime(df_pop['date'])
df_inc['date'] = pd.to_datetime(df_inc['date'])
df_pov['date'] = pd.to_datetime(df_pov['date'])

# Filter for years 2020 to 2024 (inclusive)
df_pop_filtered = df_pop[(df_pop['date'].dt.year >= 2020) & (df_pop['date'].dt.year <= 2024)]
df_inc_filtered = df_inc[(df_inc['date'].dt.year >= 2020) & (df_inc['date'].dt.year <= 2024)]
df_pov_filtered = df_pov[(df_pov['date'].dt.year >= 2020) & (df_pov['date'].dt.year <= 2024)]

# Print shapes and display previews
print(f"df_pop (2020-2024) shape: {df_pop_filtered.shape}")
print(f"df_inc (2020-2024) shape: {df_inc_filtered.shape}")
print(f"df_pov (2020-2024) shape: {df_pov_filtered.shape}")

print("\nPreview of df_pop_filtered:")
display(df_pop_filtered.head())

print("\nPreview of df_inc_filtered:")
display(df_inc_filtered.head())

print("\nPreview of df_pov_filtered:")
display(df_pov_filtered.head())
```

```python
import pandas as pd

# Filter df_pop_filtered for sex='both', age='overall', and ethnicity='overall'
df_pop_demog_filtered = df_pop_filtered[
    (df_pop_filtered['sex'] == 'both') &
    (df_pop_filtered['age'] == 'overall') &
    (df_pop_filtered['ethnicity'] == 'overall')
]

# Merge df_pop_demog_filtered with df_inc_filtered on date, state, and parlimen
df_merged = pd.merge(
    df_pop_demog_filtered,
    df_inc_filtered,
    on=['date', 'state', 'parlimen'],
    how='inner'
)

# Merge the result with df_pov_filtered on date, state, and parlimen
df_final_merged = pd.merge(
    df_merged,
    df_pov_filtered,
    on=['date', 'state', 'parlimen'],
    how='inner'
)

# Display the final merged dataframe shape and preview
print(f"Merged DataFrame shape: {df_final_merged.shape}")
display(df_final_merged.head())
```

```python

import pandas as pd
import plotly.express as px

# Filter data for Selangor
df_selangor = df_final_merged[
    df_final_merged['state'] == 'Selangor'
].copy()

# Convert date column to datetime
df_selangor['date'] = pd.to_datetime(df_selangor['date'])

# Get the latest available year
latest_year = df_selangor['date'].dt.year.max()

# Filter data for the latest year
df_latest = df_selangor[
    df_selangor['date'].dt.year == latest_year
].copy()

# Sort median household income in ascending order
df_latest = df_latest.sort_values(
    by='income_median',
    ascending=True
)

# Create horizontal bar chart
fig = px.bar(
    df_latest,
    x='income_median',
    y='parlimen',
    orientation='h',
    color='income_median',
    color_continuous_scale='Viridis',
    text='income_median',
    title=f'Median Household Income Across Parliamentary Constituencies in Selangor ({latest_year})',
    labels={
        'income_median': 'Median Household Income (RM)',
        'parlimen': 'Parliamentary Constituency'
    },
    hover_data={
        'income_median': ':,.0f',
        'income_mean': ':,.0f',
        'poverty_absolute': ':.2f'
    }
)

# Format bar labels
fig.update_traces(
    texttemplate='RM %{x:,.0f}',
    textposition='outside',
    cliponaxis=False
)

# Customize chart layout
fig.update_layout(
    template='plotly_white',
    height=800,
    coloraxis_showscale=False,
    showlegend=False,
    margin=dict(l=150, r=100, t=80, b=60),
    xaxis_title='Median Household Income (RM)',
    yaxis_title='Parliamentary Constituency',
    title_x=0.5
)

# Display chart
fig.show()
```

```python

import pandas as pd
import plotly.express as px

# Filter data for Selangor
df_selangor = df_final_merged[
    df_final_merged['state'] == 'Selangor'
].copy()

# Convert date column to datetime
df_selangor['date'] = pd.to_datetime(df_selangor['date'])

# Extract year
df_selangor['year'] = df_selangor['date'].dt.year.astype(str)

# Keep only 2022 and 2024
df_selangor = df_selangor[
    df_selangor['year'].isin(['2022', '2024'])
].copy()

# Sort constituencies by mean income in 2024
sort_order = (
    df_selangor[df_selangor['year'] == '2024']
    .sort_values(by='income_mean', ascending=False)
    ['parlimen']
    .tolist()
)

# Create interactive clustered horizontal bar chart
fig = px.bar(
    df_selangor,
    x='income_mean',
    y='parlimen',
    color='year',
    barmode='group',
    orientation='h',
    text='income_mean',
    category_orders={
        'parlimen': sort_order,
        'year': ['2022', '2024']
    },
    color_discrete_sequence=['#9ECAE1', '#2171B5'],
    title='Comparison of Mean Household Income Across Parliamentary Constituencies in Selangor (2022 vs 2024)',
    labels={
        'income_mean': 'Mean Household Income (RM)',
        'parlimen': 'Parliamentary Constituency',
        'year': 'Year'
    },
    hover_data={
        'income_mean': ':,.0f',
        'income_median': ':,.0f',
        'poverty_absolute': ':.2f'
    }
)

# Display income values on bars
fig.update_traces(
    texttemplate='RM %{x:,.0f}',
    textposition='outside',
    cliponaxis=False
)

# Customize chart layout
fig.update_layout(
    template='plotly_white',
    title_x=0.5,
    height=850,
    barmode='group',
    bargap=0.25,
    margin=dict(l=150, r=100, t=100, b=60),
    xaxis_title='Mean Household Income (RM)',
    yaxis_title='Parliamentary Constituency',
    legend_title='Year',
    hoverlabel=dict(
        bgcolor='white',
        font_size=12
    ),
    xaxis=dict(
        showgrid=True,
        gridcolor='lightgrey'
    )
)

# Display interactive chart
fig.show()
```
